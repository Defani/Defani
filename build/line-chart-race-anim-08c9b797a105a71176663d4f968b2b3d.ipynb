1. Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.animation as animation
```

2. Load Data

```python
df = pd.read_csv("Rangkaian waktu Wilayah • Wilayah berdasarkan kelas • 1990 - 2024.csv")
df_filtered = df[df['Tingkat 2'].isin(['Formasi Hutan', 'Sawit'])].copy()
```

3. Set Periode

```python
years = ['2000', '2005', '2010', '2015', '2020', '2024']
```

4. Persiapkan data biar tahun menjadi index (Sumbu X)


```python
data = df_filtered.set_index('Tingkat 2')[years].astype(float).T
data.index = pd.to_numeric(data.index)
```

8. Label teks luas di ujung garis


```python
text_hutan = ax.text(0, 0, '', color=colors['Formasi Hutan'], fontweight='bold', va='bottom', ha='right')
text_sawit = ax.text(0, 0, '', color=colors['Sawit'], fontweight='bold', va='top', ha='right')

ax.legend(loc='center right')
ax.grid(True, linestyle='--', alpha=0.5)
```

9. Memunculkan data tahun per tahun untuk animasi


```python
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

def update(frame):
    curr_years = data.index[:frame+1]

    line_hutan.set_data(curr_years, data.loc[curr_years, 'Formasi Hutan'])
    line_sawit.set_data(curr_years, data.loc[curr_years, 'Sawit'])

    if len(curr_years) > 0:
        last_year = curr_years[-1]
        val_h = data.loc[last_year, 'Formasi Hutan']
        val_s = data.loc[last_year, 'Sawit']

        text_hutan.set_position((last_year, val_h + 1000))
        text_hutan.set_text(f'{val_h:,.0f}')

        text_sawit.set_position((last_year, val_s - 2000))
        text_sawit.set_text(f'{val_s:,.0f}')

    return line_hutan, line_sawit, text_hutan, text_sawit

```
